```python
def insert_a(sin_text_list, mapped_text_list):
	sin_text_list.append("_") # Append a dummy non effecting symbol at end of sin_text_list to overcome error of length mis match of occuring in rule 2)
	a_inserted_list = []

	for i, letter in enumerate(mapped_text_list):
		a_inserted_list.append(letter)

		# rule 1 : dont insert /a/ after a wovel phoneme representative
		if sin_text_list[i] in big_list:
			continue

		# rule 2 : don't insert /a/ after a phoneme symbol, if the next relevent symbol in sin_text_list is in sripali_list.
		elif sin_text_list[i+1] in sripali_list:
			if sin_text_list[i+1] == u"ං": #to handle special cases like පංචිකාවත්ත where ං is just after a consonent
				a_inserted_list.append('a')
			continue

		# handdle spaces, commas and new line characters
		elif letter in special_character_list:
				continue

		# rule 3 : if the letter is not filtered by above rules, then append /a/
		else:
			a_inserted_list.append('a')

	return a_inserted_list
```

```python
	
def transliterate(self):
    sin_text = self

    #remove english and unwanted symbols
    sin_text_temp_list = list(sin_text)
    sin_text_list = [x for x in sin_text_temp_list if x not in unwanted_symbols]
    #print("original is: ", sin_text_temp_list)
    #print("unwanted symbols removed list is ", sin_text_list)
    sin_text = "".join(sin_text_list)
    #print("unwanted symbols removed sin_text is ", sin_text)
    #print(type(sin_text))


    #call convert_to_phoneme function and get the halfly transliterated string
    mapped_text_list = convert_to_phoneme(sin_text)
    #print("mapped text list: ",mapped_text_list)


    a_inserted_list = insert_a(sin_text_list, mapped_text_list)
    #print( "a inserted phoneme string:", a_inserted_list)
    #print("\n")

    singlish_list = phoneme_to_english(a_inserted_list)
    #print("singlish list", singlish_list)
    singlish_word=""
    for i in singlish_list:
        singlish_word+=i
    #print(singlish_word)
    return singlish_word
```

```python
import os
import pandas as pd

def remove_commas(text):
    """Removes commas from the given text."""
    return text.replace(",", "")

def find_word_positions(sentence, words):
    """
    Finds the positions of specified words in a sentence.
    Returns a sorted list of tuples (word, position) by occurrence order.
    """
    tokens = sentence.split()
    positions = []
    for word in words:
        if word in tokens:
            positions.append((word, tokens.index(word) + 1))  # +1 for 1-based indexing
    return sorted(positions, key=lambda x: x[1])

def replace_with_filename(sentence, positions, filename):
    """
    Replaces words at specified positions with the filename (without extension).
    Returns modified sentence and replaced words.
    """
    tokens = sentence.split()
    replaced_words = []
    
    for word, pos in positions:
        if pos <= len(tokens):
            replaced_words.append(tokens[pos-1])
            tokens[pos-1] = os.path.splitext(filename)[0]  # Remove file extension
    
    return ' '.join(tokens), ', '.join(replaced_words)

def process_files(input_dir, output_excel):
    # Initialize DataFrame storage
    data = []
    
    # Process all files in input directory
    for filename in os.listdir(input_dir):
        if filename.endswith('.txt'):
            input_path = os.path.join(input_dir, filename)
            
            with open(input_path, 'r', encoding='utf-8') as infile:
                lines = [line.strip() for line in infile if line.strip()]
            
            current_keywords = []
            current_section = []
            
            for line in lines:
                if line.startswith('#'):
                    # Process previous section
                    if current_section:
                        for sentence in current_section:
                            clean_sentence = remove_commas(sentence)
                            romanized = transliterate(clean_sentence)
                            
                            # Get ordered ambiguity words
                            positions = find_word_positions(clean_sentence, current_keywords)
                            ambiguity_words = ', '.join([word for word, _ in positions])
                            
                            # Replace ambiguity words with filename in Romanized version
                            modified_romanized, replaced_words = replace_with_filename(
                                romanized,
                                positions,
                                filename
                            )
                            
                            data.append({
                                'Romanized Sinhala': modified_romanized,
                                'Native Sinhala': clean_sentence,
                                'Ambiguity Word(s)': ambiguity_words,
                                'Romanized Ambiguity Word': replaced_words,
                                'Source File': os.path.splitext(filename)[0]
                            })
                    
                    # Update current keywords
                    current_keywords = [w.strip() for w in line[1:].split('-')]
                    current_section = []
                else:
                    current_section.append(line)
            
            # Process remaining sentences
            if current_section:
                for sentence in current_section:
                    clean_sentence = remove_commas(sentence)
                    romanized = transliterate(clean_sentence)
                    
                    positions = find_word_positions(clean_sentence, current_keywords)
                    ambiguity_words = ', '.join([word for word, _ in positions])
                    
                    modified_romanized, replaced_words = replace_with_filename(
                        romanized,
                        positions,
                        filename
                    )
                    
                    data.append({
                        'Romanized Sinhala': modified_romanized,
                        'Ambiguity Word': clean_sentence,
                        'Sinhala Words': ambiguity_words,
                        'Romanized Ambiguity Word': replaced_words,
                        'Source File': os.path.splitext(filename)[0]
                    })
    
    # Create DataFrame and save to Excel
    df = pd.DataFrame(data)
    df.to_excel(output_excel, index=False, engine='openpyxl')

# Example usage
input_directory = "E:/4th Year/FYP/FYP-TEST/Sentences/sinhala_text_before_transliteration"
output_directory = "E:/4th Year/FYP/FYP-TEST/Sentences/excel/transliterated_sentences.xlsx"
# Example usage
process_files(input_directory, output_directory)
```
