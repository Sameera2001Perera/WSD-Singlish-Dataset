```python
import getpass
import os
from langchain_core.prompts import ChatPromptTemplate
from langchain_anthropic import ChatAnthropic


if "ANTHROPIC_API_KEY" not in os.environ:
    # os.environ["ANTHROPIC_API_KEY"] = getpass.getpass("Enter your Anthropic API key: ")
    os.environ["ANTHROPIC_API_KEY"] = "###################"
```

```python
llm = ChatAnthropic(
    model="claude-3-7-sonnet-20250219",
    temperature=0.2,
    timeout=None,
    max_retries=2,
    max_tokens=5000
)
```

```python
input_text = """A back-transliteration system has been developed to convert Romanized Sinhala text into the native Sinhala script. This system accurately handles Romanized Sinhala words that correspond to multiple Sinhala words by considering contextual information.

To evaluate this transliteration system, a test dataset needs to be created. I want to generate Sinhala sentences for specific ambiguous words.

The ambiguous Romanized Sinhala word in focus is {singlish_word}, which can correspond to either {sinhala_word1} or {sinhala_word2} in Sinhala.

Below are the example sentences that include only one ambiguous word.

Sentences that include {sinhala_word1}:
{sentences1}

Sentences that include {sinhala_word2}:
{sentences2}

Considering the context, generate 20 Sinhala sentences that include both {sinhala_word1} and {sinhala_word2} in the same sentence.

Ensure that the sentences provide sufficient contextual information to evaluate how well the transliterator resolves ambiguity based on context."""
```

```python

prompt = ChatPromptTemplate(
    [
        (
            "system",
            "You are a helpful assistant specialized in generating natural sounding Sinhala sentences.",
        ),
        ("human", input_text),
    ]
)

chain = prompt | llm
```
